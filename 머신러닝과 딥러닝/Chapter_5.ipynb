>> Linear Transform (선형 변환) : 벡터에 행렬을 곱하여 벡터의 차원과 값이 함꼐 바뀌는 것

```python
#[실습 05-08]

#벡터의 선형 변환(1)
x = np.array([1,2])

W1 = np.array([[1,0,0.5],
               [0,1,0.5]])

x1 = x @ W1

print(f"x = {x}")
print("==== W1 선형 변환 후 ====")
print(f"x1 = {x1}")
```

```python
#[실습 05-09]

#벡터의 선형 변환(2)

W2 = np.array([[0.3,0.7,0.1,-0.2],
               [0.3,0.2,0.3,0.1],
               [0.2,0.4,-0.2,0.2]])

x2 = x1 @ W2

print(f"x1 = {x1}")
print("==== W2 선형 변환 후 ====")
print(f"x2 = {x2}")
```

>> k차원 벡터에 k x q 행렬을 곱하면 q차원 벡터로 변환된다

>> 즉, x2를 다시 선형변환하여 하나의 스칼라를 얻으려면 x2에 4 x 1 행렬을 곱하면 된다
>> (2차원 벡터 x에 가중치행렬 w1을 곱해서 3차원 벡터 x1을 만들고 x1에 가중치 행렬 w2을 곱해서 4차원 벡터 x2를 만든 상태)

```python
#[실습 05-10]

#벡터의 선형 변환(3)

W3 = np.array([[0.1],
               [0.3,],
               [0.5],
               [-0.2]])

x3 = x2 @ W3

print(f"x2 = {x2}")
print("==== W3 선형 변환 후 ====")
print(f"x3 = {x3}")
```

```python
#[실습 05-11]

#입력 행렬(3x2)
X = np.array([[1,2],
              [3,4],
              [5,6]])

#행렬 곱으로 선형 변환 반복
X1 = X @ W1         #1차 변환
X2 = X1 @ W2        #2차 변환
X3 = X2 @ W3        #3차 변환

print("1차 변환 결과(X1)")
print(X1)
print()

print("2차 변환 결과(X1)")
print(X2)
print()

print("3차 변환 결과(X1)")
print(X3)
```

>> ANN(인공신경망)에서는 모든 데이터를 Tensor이라는 자료형으로 변환해서 사용
>> Tensor는 배열과 동일한 형태
>> 고속 행렬 연산과 자동 미분 기능 지원

>> Gradient : 모든 변수에 대한 미분을 모아 놓은 벡터
>> Gradient는 머신러닝에서 파라미터를 최적화하기 위한 방향을 알려준다
>> (모델의 학습 = 손실을 가장 작게 하는 파라미터 찾기)
>> 손실을 줄이려면 Gradient 부호의 반대 방향으로 파라미터 변환
>> 손실을 빠르게 줄이려면 Gradient의 절댓값이 큰 파라미터를 더 큰 폭으로 업데이트
    W_new = W_old - G(loss,W_old)
    >> Overshooting 발생 가능성 존재 (너무 많이 이동해 손실의 최솟값을 지나침)

>> 학습률(learning rate) : 0보다는 큰 작은 값, Gradient의 변화의 크기를 조절하는 역할
>> 경사하강법 : 파라미터 최적화 방법 중 아래 공식 사용(학습률 반영)
    W_new = W_old - ⍺ x G(loss, w_old)



>> 자동 미분 (Automatic Differentiation) :  사용자가 특정 변수의 그래디언트를 요청하면 텐서가 기록된 연산 과정을 이용해 필요한 미분값을 자동으로 계산하는 기능
>> Tensorflow가 자동 미분 기능을 제공

```python
#[실습 05-13]


!pip install koreanize_matplotlib

import koreanize_matplotlib
import numpy as np
import tensorflow as tf
import matplotlib.pyplot as plt

np.set_printoptions(precision=5, suppress=True)

# 텐서 변수 정의
x_data = tf.cast(X, tf.float32) #tf.cast는 numpy 배열을 Tensor로 변환하는 함수 (X to x_data)
y_data = tf.cast(y, tf.float32)

# 가중치 행렬w, 편향 b를 무작위 값으로 초기화
w = tf.Variable(tf.random.normal([2,1])) # tf.random.normal : 정규분포를 따르는 난수로 텐서 생성
b = tf.Variable(tf.random.normal([1,]))  # tf.Variable() : 값이 계속 변경되는 텐서 변수를 지정

learning_rate = 0.05    # 학습률

# 결과 기록 리스트 초기화
w_history = []
b_history = []
loss_history = []

# 학습 루프
for i in range(500):            # 파라미터를 총 500회 업데이트 (500회 스텝 수행)
    # 자동 미분을 위한 연산 기록
    with tf.GradientTape() as tape: # 연산 과정을 테이프처럼 기록
        # 순전파
        y_pred = tf.matmul(x_data, w) + b   # 입력 데이터와 가중치 행렬의 곱
        # MSE 손실
        loss = tf.reduce_mean(tf.square(y_pred - y_data))   #reduce_mean : 평균 계산 함수

    # 그래디언트 계산
    dw, db = tape.gradient(loss, [w,b]) # 미분해서 그래디언트 계산

    # 경사하강법으로 파라미터 업데이트
    w.assign_sub(learning_rate * dw)
    b.assign_sub(learning_rate * db)

    # 결과 저장
    w_history.append(w.numpy().flatten())   # 현재 파라미터와 손실을 numpy() 메서드로 넘파이 배열 형태로 변환하여 저장
    b_history.append(b.numpy().flatten())
    loss_history.append(loss.numpy())

print(f"=== 학습 완료 ===")
print(f"최종 가중치 w:\n{w.numpy()}")   
print()
print(f"최종 절편 b:{b.numpy()[0]:.3f}")
print()
print(f"최종 손실 :{loss.numpy():.3f}")

plt.plot(loss_history)
plt.title("스텝별 손실 변화")
plt.xlabel("스텝")
plt.ylabel("평균제곱오차")
plt.grid(True)

plt.show()
```

>> 순전파(Forward Propagation) : 입력 데이터 / 현재 파라미터를 사용해 타깃의 예측값과 손실을 계산 (행렬연산)
>> 역전파(Backward Propagation) : 그래디언트를 구해 이를 경사하강법으로 파라미터를 업데이트 (미분)

```python
# 되새김_5장_Q5_Learning_rate_학습

import koreanize_matplotlib

import matplotlib.pyplot as plt
import numpy as np
import tensorflow as tf
import pandas as pd


def draw_45(y,y_pred,mse):

    plt.figure(figsize=(8.6, 6))

    plt.scatter(y,y_pred)

    plt.plot(
        [y.min(), y.max()],
        [y.min(),y.max()],
        color='red',
        linestyle='--',
        label='45° Line'
    )

    plt.title(f'예측값과 실젯값(MSE = {mse:.3f}')
    plt.xlabel('실젯값')
    plt.ylabel('예측값')
    plt.legend()
    plt.show()


# 데이터 불러오기

df = pd.read_csv("혈압.csv", encoding='cp949')

y = np.array(df['혈압']).reshape(1000,1)

X = np.array(
    df[['나이','체중','키','수면시간']]
).reshape(1000,4)


# 텐서 변수 정의

x_data = tf.cast(X, tf.float32)
y_data = tf.cast(y, tf.float32)


# 가중치 초기화

tf.random.set_seed(42)

w1 = tf.Variable(tf.random.normal([4,2]))
w2 = tf.Variable(tf.random.normal([2,1]))

b1 = tf.Variable(tf.random.normal([1,1]))
b2 = tf.Variable(tf.random.normal([1,1]))


# 처음에는 크게 설정
learning_rate = 0.01


for i in range(1000):

    # 현재 MSE 계산
    with tf.GradientTape() as tape:

        x_pred = tf.matmul(x_data, w1) + b1
        y_pred = tf.matmul(x_pred, w2) + b2

        mse = tf.reduce_mean(
            tf.square(y_data - y_pred)
        )

    dw1, dw2, db1, db2 = tape.gradient(
        mse, [w1, w2, b1, b2]
    )

    current_mse = mse.numpy()


    # 현재 가중치 저장
    old_w1 = w1.numpy().copy()
    old_w2 = w2.numpy().copy()
    old_b1 = b1.numpy().copy()
    old_b2 = b2.numpy().copy()


    # learning rate가 적절할 때까지 자동으로 조절
    while True:

        # 가중치 업데이트
        w1.assign_sub(learning_rate * dw1)
        w2.assign_sub(learning_rate * dw2)
        b1.assign_sub(learning_rate * db1)
        b2.assign_sub(learning_rate * db2)


        # 업데이트 후 MSE 계산
        new_x_pred = tf.matmul(x_data, w1) + b1
        new_y_pred = tf.matmul(new_x_pred, w2) + b2

        new_mse = tf.reduce_mean(
            tf.square(y_data - new_y_pred)
        )

        new_mse_value = new_mse.numpy()


        # MSE가 정상적으로 감소했으면 업데이트 확정
        if np.isfinite(new_mse_value) and new_mse_value < current_mse:

            x_pred = new_x_pred
            y_pred = new_y_pred
            mse = new_mse

            break


        # MSE가 증가하거나 nan이면 이전 값으로 복구
        w1.assign(old_w1)
        w2.assign(old_w2)
        b1.assign(old_b1)
        b2.assign(old_b2)

        # learning rate를 절반으로 줄임
        learning_rate *= 0.5


    if i % 100 == 0:

        print(
            i,
            mse.numpy(),
            "learning_rate =",
            learning_rate
        )


    if mse<100:
        draw_45(y,y_pred,mse)
        break
```

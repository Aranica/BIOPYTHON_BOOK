```python
#[실습 02-35]

import pandas as pd
import numpy as np

np.random.seed(42)

week=['일','월','화','수','목','금','토']
weeks = week * 5

print(weeks[:5])
print()

sales = np.random.randint(500,1000,35)

#딕셔너리 형태로 변경
data={'요일':weeks,'판매량':sales}

#데이터프레임 변환
df = pd.DataFrame(data)
df[:5]
```

```python
#[실습 02-36]

요일_평균 = []

for i in range(7):
    df1 = df[df["요일"]==week[i]];
    z = df1["판매량"].mean()
    요일_평균.append(z)

print(요일_평균)
print()

pd.DataFrame(요일_평균,index=week, columns=["평균_판매량"])
```

```python
#[실습 02-37]

#판매량의 전체 평균 계산
m = df["판매량"].mean()
print("판매량_전체_평균 =",m)

#조건이 1개인 경우
for i in range(7):
    avg = 요일_평균[i]
    if avg >= m+50:
        print(f"{week[i]}요일은 판매량이 매우 높습니다! (평균 {avg:.1f})")
```

```python
#[실습 02-38]

for i in range(7):
    avg = 요일_평균[i]

    if avg >= m+50 :
        print(f"{week[i]}요일은 판매량이 매우 높습니다! (평균 {avg:.1f})")

    elif avg >= m-50 :
            print(f"{week[i]}요일은 판매량이 보통 수준입니다. (평균 {avg:.1f})")

    else :
            print(f"{week[i]}요일은 판매량이 낮습니다 (평균 {avg:.1f})")
```

```python
#되새김_2장_Q1
#연도가 2010년 이후인 세단형 자동차만 골라 브랜드별 평균 판매 가격을 막대그래프로 표시

import matplotlib.pyplot as plt

car=pd.read_csv("car.csv",encoding='cp949')

# 1.세단만 스타일에서 골라 sedan에 넣는다
# 2.sedan에서 연도가 2010년 이후인 집합을 sedan1으로 모은다
# 3. sedan1에서 평균 가격을 설정하여 막대그래프로 표시한다
# 4. ㄱㄱ

sedan = car[car["스타일"]=="세단"]
sedan1 = sedan[sedan["연도"]>=2010]
sedan1.groupby("브랜드")["가격"].mean().plot.bar()

plt.show()
```
